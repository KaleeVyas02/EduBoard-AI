## 1. Import libraries

```python
from pathlib import Path
import cv2
import matplotlib.pyplot as plt
from pipeline import read_image, image_quality, correct_perspective, enhance_image, extract_text
```

## 2. Select one photograph

```python
photos = sorted(Path('sample_images').glob('*.jpg'))
print('Photos found:', len(photos))
image = read_image(photos[0])
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off'); plt.show()
```

## 3. Check image quality

```python
print('Original quality:', image_quality(image))
```

## 4. Perspective correction

```python
corrected, board_found = correct_perspective(image)
print('Board automatically detected:', board_found)
plt.imshow(cv2.cvtColor(corrected, cv2.COLOR_BGR2RGB))
plt.axis('off'); plt.show()
```

## 5. Enhance contrast and sharpness

```python
enhanced = enhance_image(corrected)
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1); plt.imshow(cv2.cvtColor(corrected, cv2.COLOR_BGR2RGB)); plt.title('Before'); plt.axis('off')
plt.subplot(1, 2, 2); plt.imshow(cv2.cvtColor(enhanced, cv2.COLOR_BGR2RGB)); plt.title('After'); plt.axis('off')
plt.show()
```

## 6. Extract text using pretrained OCR

```python
original_text = extract_text(corrected)
enhanced_text = extract_text(enhanced)
print('Before enhancement:\n', original_text[:2000])
print('\nAfter enhancement:\n', enhanced_text[:2000])
```

## 7. Save enhanced image and notes

```python
Path('notebook_output').mkdir(exist_ok=True)
cv2.imwrite('notebook_output/enhanced.jpg', enhanced)
Path('notebook_output/notes.txt').write_text(enhanced_text, encoding='utf-8')
print('Saved files in notebook_output folder')
```
